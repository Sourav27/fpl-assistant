```python
pd.options.display.max_columns = None
```

```python
url = 'https://fantasy.premierleague.com/api/bootstrap-static/'
```

```python
r = requests.get(url)
```

```python
json = r.json()
```

```python
elements_df = pd.DataFrame(json['elements'])
```

```python
for x in elements_df.index :
    print(x)
    element_id = elements_df.id[x]
    url = f'https://fantasy.premierleague.com/api/element-summary/{element_id}/'
    r = requests.get(url)
    json = r.json()
    json_history_df = pd.DataFrame(json['history'])
    json_history_past_df = pd.DataFrame(json['history_past'])
    
    if x == 0 :
        all_history_df = json_history_df
        all_history_past_df = json_history_past_df
    else : 
        all_history_df = pd.concat([all_history_df, json_history_df])
        all_history_past_df = pd.concat([all_history_past_df, json_history_past_df])
        
    time.sleep(3)
```

```python
all_history_df.to_csv('all_history_df_2223.csv')
all_history_past_df.to_csv('all_history_past_df_2223.csv')
```
