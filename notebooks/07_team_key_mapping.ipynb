```python
import numpy as np
import pandas as pd
```

```python
merged_data = pd.read_csv(r"D:\FPL\fpl-optimization\point-predictor\code\xP Model\data\all_players_timeseries_dataset.csv", index_col=False)
```

```python
# Extract unique values from 'team_x' column and sort them
unique_teams = pd.Series(merged_data['team_x'].unique()).sort_values().reset_index(drop=True)

# Create a DataFrame with a 'team_code' column and a 'team' column
unique_teams_df = pd.DataFrame({
    'team_code': unique_teams.index + 1,  # Create a sequential team code starting from 1
    'team': unique_teams
})

# Display the resulting DataFrame
unique_teams_df
```

```python
unique_teams_df.to_csv(r'data/team_key.csv', index=False)
```
